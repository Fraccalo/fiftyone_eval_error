# Evaluation error on image 

```python
import fiftyone as fo
from fiftyone.utils.coco import add_coco_labels
```

```python
dataset_name = "test_error"
image_folder = ""
```

```python
if fo.dataset_exists(dataset_name):
    fo.delete_dataset(dataset_name)
```

```python
overwrite_dataset = True
annotation_file = "gt_error.json"
annotation_label = "gt"
label_types = ['detections',"segmentations"]


dataset = fo.Dataset.from_dir(data_path = image_folder,
                    labels_path = annotation_file,
                    dataset_type = fo.types.COCODetectionDataset,
                    name = dataset_name,
                    label_types = label_types,
                    label_field = annotation_label)
```

```python
model_label = "preds"
prediction_file = "preds_error.json"
label_types = "segmentations"

add_coco_labels(sample_collection = dataset,
                label_field = model_label,
                labels_or_path = prediction_file,
                # classes = {i:str(i) for i in range(1,21)}, # use for version 0.25.2,
                categories = {i:str(i) for i in range(1,21)}, # use for version 1.3,
                label_type = label_types)
```

```python
compute_mAP = True
classwise = False
use_masks = True
model_label = "preds"
annotation_label = "gt_segmentations"
evaluation_label = "evaluation"

dataset.evaluate_detections(pred_field = model_label,
                            gt_field = annotation_label,
                            eval_key = evaluation_label,
                            compute_mAP = compute_mAP,
                            classwise = classwise,
                            use_masks = use_masks)
```
